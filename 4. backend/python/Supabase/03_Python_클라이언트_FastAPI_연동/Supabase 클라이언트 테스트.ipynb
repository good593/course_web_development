# 라이브러리 임포트

```python
import os
from dotenv import load_dotenv
from supabase import create_client, Client
```

# .env 파일의 데이터 환경변수에 등록 

```python
load_dotenv()
```

# 환경변수에 등록된 데이터 변수에 저장 

```python
SUPABASE_URL = os.environ["SUPABASE_URL"]
SUPABASE_SERVICE_ROLE_KEY = os.environ["SUPABASE_SERVICE_ROLE_KEY"]
```

# Supabase 클라이언트 생성 

```python
supabase: Client = create_client(SUPABASE_URL, SUPABASE_SERVICE_ROLE_KEY)
```

# 간단한 조회로 연결 테스트

```python
result = supabase.table("app_users").select("*").limit(1).execute()
print(result.data)  # [] 또는 데이터 리스트 출력되면 연결 성공
```

# 데이터 조회

### 전체 조회

```python
result = supabase.table("app_users").select("*").execute()
users = result.data  # 딕셔너리 리스트
print(users)
```

```python
result = (
    supabase.table("app_users")
    .select("id, username, created_at")
    .order("created_at", desc=True)
    .execute()
)

result.data
```

```python
result = (
    supabase.table("conversations")
    .select("*")
    .eq("user_id", "11acf377-3c31-42c9-a5fd-6cd9c0d8b4a2")  # WHERE user_id = '...'
    .order("created_at", desc=True)
    .execute()
)

conversations = result.data
conversations
```
